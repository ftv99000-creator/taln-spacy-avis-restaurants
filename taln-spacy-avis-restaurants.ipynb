## Exercice 1.1 — annoter

```python
def annoter(doc):
    lignes = []
    for t in doc:
        lignes.append({"token": t.text, "lemme": t.lemma_, "pos": t.pos_, "traits": str(t.morph)})
    return pd.DataFrame(lignes)

annoter(docs[0])
```

```python
# ✅ Vérification (ne pas modifier)
_a = annoter(nlp("Les serveurs sont aimables."))
verifier("4 colonnes", lambda: list(_a.columns) == ["token", "lemme", "pos", "traits"])
verifier("une ligne par token", lambda: len(_a) == 5 and _a.loc[1, "lemme"] == "serveur" and _a.loc[1, "pos"] == "NOUN")
```

## Exercice 1.2 — catégories du corpus

```python
freq_pos = Counter(t.pos_ for d in docs for t in d)
print(freq_pos.most_common())
etiq, nb = zip(*freq_pos.most_common())
plt.figure(figsize=(8, 3)); plt.bar(etiq, nb, color="#0B8E36"); plt.title("Catégories (UD) du corpus"); plt.show()
```

```python
# ✅ Vérification (ne pas modifier)
verifier("freq_pos", lambda: freq_pos == Counter(t.pos_ for d in docs for t in d))
```

## Exercice 1.3 — adjectifs par classe

```python
def adjectifs(docs, etiquettes, classe):
    return Counter(t.lemma_.lower() for d, e in zip(docs, etiquettes) if e == classe
                   for t in d if t.pos_ == "ADJ")

print("positifs :", adjectifs(docs, etiquettes, "pos").most_common(8))
print("négatifs :", adjectifs(docs, etiquettes, "neg").most_common(8))
```

```python
# ✅ Vérification (ne pas modifier)
_d = list(nlp.pipe(["Un plat délicieux.", "Un service lent."]))
verifier("adjectifs par classe", lambda: adjectifs(_d, ["pos", "neg"], "pos") == Counter({"délicieux": 1}) and adjectifs(_d, ["pos", "neg"], "neg") == Counter({"lent": 1}))
```

## 💬 Interprétation 1
**(a)** Les trois catégories les plus fréquentes sont **NOUN (263), PUNCT (199) et ADJ (164)**, suivies de DET (124) et ADV (118). Il y a beaucoup de **PUNCT** car les avis sont courts, expressifs et ponctués (« ! », « . », « , », « : ») ; il y a beaucoup d'**ADJ** car un avis sert à *évaluer* : « délicieux », « rapide », « froid », « lent »…

**(b)** Les adjectifs dominants sont ceux attendus : positifs → *bon* (20), *rapide* (15), *délicieux* (12), *chaleureux*, *magnifique*, *agréable*, *excellent* ; négatifs → *froid* (12), *lent* (6), *sale*, *élevé*. Erreurs observées : *salé/salée* est lemmatisé en **« saler »** (le verbe) dans les avis négatifs ; dans « trop salé », *salé* est étiqueté **ADV** ; dans « du tout », *tout* est étiqueté NOUN ; les mots mal orthographiés (« magnifque » → lemme *magnifqu*) sont mal lemmatisés. **« bon » apparaît 4 fois dans les avis négatifs** parce que la négation est portée par un autre mot : « n'était **pas bon** », « pas bon, pas chaud » — de même *rapide* apparaît 3 fois côté négatif (« pas rapide », « livraison rapide mais… »). L'étiquetage POS ne voit pas la polarité : il compte *bon* comme ADJ, que le contexte soit positif ou négatif.

## Exercice 2.2 — BIO

```python
def bio(doc):
    sortie = []
    for t in doc:
        etiquette = "O" if t.ent_iob_ == "O" else t.ent_iob_ + "-" + t.ent_type_
        sortie.append(f"{t.text}/{etiquette}")
    return sortie

print(" ".join(bio(docs_ner[4])))
```

```python
# ✅ Vérification (ne pas modifier)
verifier("BIO", lambda: bio(docs_ner[4])[:4] == ["Le/O", "chef/O", "Karim/B-PER", "Benali/I-PER"])
```

## Exercice 2.3 — extracteur d'entités (livrable)

```python
def extraire_entites(docs):
    lignes = []
    for i, doc in enumerate(docs):
        for texte, typ in entites(doc):
            lignes.append({"avis": i, "entité": texte, "type": typ})
    return pd.DataFrame(lignes, columns=["avis", "entité", "type"])

table_entites = extraire_entites(docs_ner)
print(table_entites["type"].value_counts().to_dict())
table_entites
```

```python
# ✅ Vérification (ne pas modifier)
verifier("colonnes", lambda: list(table_entites.columns) == ["avis", "entité", "type"])
verifier("une ligne par entité", lambda: len(table_entites) == sum(len(d.ents) for d in docs_ner))
```

## 2.4 Évaluer le modèle (fourni)

```python
justes = trouvees = attendues = 0
for i, (doc, gold) in enumerate(zip(docs_ner, GOLD)):
    pred = set(entites(doc))
    justes += len(pred & gold); trouvees += len(pred); attendues += len(gold)
    if pred != gold:
        print(f"avis {i:2} | manquées : {sorted(gold - pred)} | en trop : {sorted(pred - gold)}")
print(f"\nPrécision = {justes}/{trouvees} = {justes / trouvees:.2f} | Rappel = {justes}/{attendues} = {justes / attendues:.2f}")
```

## Exercice 3.2 — qui fait quoi

```python
def qui_fait_quoi(doc):
    triplets = []
    for t in doc:
        if t.pos_ == "VERB":
            sujets = [e for e in t.children if e.dep_.startswith("nsubj")]
            objets = [e for e in t.children if e.dep_ == "obj"]
            if sujets and objets:
                triplets.append((sujets[0].text, t.lemma_, objets[0].text))
    return triplets

print(qui_fait_quoi(nlp("Le chef prépare un tajine et les clients adorent le couscous.")))
print([x for d in docs for x in qui_fait_quoi(d)][:10])
```

```python
# ✅ Vérification (ne pas modifier)
verifier("qui fait quoi", lambda: qui_fait_quoi(nlp("Le chef prépare un tajine.")) == [("chef", "préparer", "tajine")])
```

## Exercice 3.3 — opinions

```python
def opinions(doc):
    paires = []
    for t in doc:
        if t.pos_ == "ADJ":
            if t.dep_ == "amod" and t.head.pos_ in ("NOUN", "PROPN"):
                paires.append((t.head.lemma_.lower(), t.lemma_.lower()))
            for e in t.children:
                if e.dep_.startswith("nsubj") and e.pos_ in ("NOUN", "PROPN"):
                    paires.append((e.lemma_.lower(), t.lemma_.lower()))
    return paires

print(opinions(nlp("Le service est lent mais le tajine est excellent.")))
couples = Counter(p for d in docs for p in opinions(d))
print(couples.most_common(12))
```

```python
# ✅ Vérification (ne pas modifier)
verifier("règle copule", lambda: ("service", "lent") in opinions(nlp("Le service est lent.")))
verifier("règle amod", lambda: ("tajine", "délicieux") in opinions(nlp("Un tajine délicieux.")))
```

## 💬 Interprétation 3
**(a)** Les couples les plus fréquents sont (service, rapide) ×14, (service, lent) ×6, (tajine, délicieux) ×4, (rapport, bon) ×4, (cadre, magnifique) ×3, (livraison, rapide) ×3, (prix, élevé) ×3, (plat, froid) ×3, (accueil, chaleureux) ×3. Ils relient chaque jugement à un **aspect précis** (service, plats, accueil, cadre, prix, livraison) : on voit *quoi* est apprécié ou critiqué, ce que les collocations du TP4 (bigrammes fréquents, simplement adjacents) ne donnent que partiellement : une collocation exige la proximité des mots, alors que la dépendance relie « Le service **était** très **lent** » malgré la distance, et distingue la cible de l'opinion. Limite : la polarité n'apparaît pas (« bon » vs « pas bon »).

**(b)** Exemples d'erreurs du modèle : dans « Le tajine n'était pas bon du tout, trop salé », *salé* est étiqueté **ADV** et *tout* **NOUN**, donc l'opinion « trop salé » est perdue ; et le couple (tajine, bon) est extrait sans la négation « pas », ce qui donne une opinion **positive à tort**. Autres cas observés : « Je ne recommande pas ce restaurant » donne ('Réservation', 'recommander', 'restaurant') et « Personnel… parfait » donne ('Personnel', 'parfaire', 'merci') en « qui fait quoi » (*parfait* étiqueté VERB) ; « Le plat est arrivé froid » ne produit aucun couple (*froid* est rattaché comme xcomp du verbe, pas comme copule) ; « Accueil chaleureux » aussi (*chaleureux* est analysé comme flat:name) ; « Service TRÈS lent » → *lent* étiqueté ADV ; « Le Café Atlas de Rabat sert… » donne le couple absurde (rabat, sert) car *sert* est étiqueté ADJ ; « livrée froide » donne (froide, livrer) ; « Le restaurent est magnifque » donne (restaurent, magnifqu), lemmes faux dus aux fautes.

## Pour aller plus loin — A. EntityRuler

```python
nlp2 = spacy.load("fr_core_news_sm")
ruler = nlp2.add_pipe("entity_ruler", before="ner")
noms = ["Dar Zitoun", "Café Atlas", "Riad Bleu", "Le Palmier", "Maison Andalouse", "TajExpress", "CIH"]
ruler.add_patterns([{"label": "ORG", "pattern": nom} for nom in noms])

justes = trouvees = attendues = 0
for doc, gold in zip(nlp2.pipe(t for t, _ in AVIS_NER), GOLD):
    pred = set(entites(doc)); justes += len(pred & gold); trouvees += len(pred); attendues += len(gold)
print(f"Précision = {justes / trouvees:.2f} | Rappel = {justes / attendues:.2f}")
```

## Pour aller plus loin — B. Négation

```python
def opinions_neg(doc):
    paires = []
    for t in doc:
        if t.pos_ != "ADJ":
            continue
        neg = any(e.dep_ == "advmod" and e.lemma_.lower() in ("pas", "jamais") for e in t.children)
        mot = ("pas " if neg else "") + t.lemma_.lower()
        if t.dep_ == "amod" and t.head.pos_ in ("NOUN", "PROPN"):
            paires.append((t.head.lemma_.lower(), mot))
        for e in t.children:
            if e.dep_.startswith("nsubj") and e.pos_ in ("NOUN", "PROPN"):
                paires.append((e.lemma_.lower(), mot))
    return paires

negatifs = Counter(p for d in docs for p in opinions_neg(d) if p[1].startswith("pas "))
print(negatifs)
```

## 💬 Interprétation A-B
**(a)** L'EntityRuler fait passer la précision de 0,36 à environ **0,73** et le rappel à environ **0,76** sur les 12 avis (valeurs à confirmer chez vous) : les noms connus sont toujours reconnus avec le bon type ORG. Limite : c'est une liste fermée — elle ne généralise pas aux restaurants absents de la liste, demande une maintenance manuelle, et peut créer des erreurs d'homonymie (ex. *Le Palmier* dans « un palmier »). Elle ne corrige pas non plus ce qui n'est pas dans la liste : *Salma*, *Fatima*, *Orange* restent mal typés, et « serveur Youssef » reste une erreur de frontière. De plus, ces règles utilisées sur le jeu d'évaluation lui-même gonflent artificiellement le score.

**(b)** On trouve 6 couples négatifs : (couscous, pas bon), (plat, pas bon), (personnel, pas aimable), (poisson, pas frais), (prix, pas élevé), (légume, pas cher). Attention : « pas cher » (et « pas élevé ») est en réalité *positif* ; il faut donc un lexique d'opinion qui traite la négation, pas une simple inversion. Sans négation, « pas bon » et « bon » donnent le même couple (plat, bon) : un avis négatif serait compté comme positif. Il faut donc inverser (ou marquer) l'opinion quand l'adjectif est gouverné par « pas » / « jamais » avant d'utiliser les couples pour estimer le sentiment.
